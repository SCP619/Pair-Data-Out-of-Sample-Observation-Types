```python
# ---------- 1. Load data ----------
# Adjust loading depending on your exact format (e.g., np.loadtxt, genfromtxt, pandas.read_csv)

X = np.loadtxt("input.data")      # shape (400, n_features)
y = np.loadtxt("output.data")     # shape (400,)
pairs = np.loadtxt("pairs.data", dtype=str)  # if you need the IDs later

print("Data shapes:", X.shape, y.shape, pairs.shape)#me 

X = StandardScaler().fit_transform(X)#me

# Ensure shapes
if y.ndim > 1 and y.shape[1] == 1:
    y = y.ravel()

n_samples = X.shape[0]
assert n_samples == y.shape[0], "X and y must have same number of samples"

# ---------- 2. Define C-index ----------

def c_index(y_true, y_pred):
    """
    Concordance index with 0.5 for ties in predictions.
    """
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    n = len(y_true)

    concordant = 0
    discordant = 0
    ties = 0

    for i in range(n):
        for j in range(i + 1, n):
            if y_true[i] == y_true[j]:
                # Often excluded, but we’ll count as a tie
                continue
            diff_true = y_true[i] - y_true[j]
            diff_pred = y_pred[i] - y_pred[j]
            prod = diff_true * diff_pred

            if prod > 0:
                concordant += 1
            elif prod < 0:
                discordant += 1
            else:
                ties += 1

    total = concordant + discordant + ties
    if total == 0:
        return np.nan  # or 0.5, depending on convention

    return (concordant + 0.5 * ties) / total

# ---------- 3. LOO cross-validation with KNN (k=10) ----------

k = 10
y_pred_all = np.zeros(n_samples)

for i in range(n_samples):
    # Indices for training (all except i)
    train_idx = np.arange(n_samples) != i

    X_train, y_train = X[train_idx], y[train_idx]
    X_test = X[~train_idx].reshape(1, -1)

    # Train KNN regressor
    knn = KNeighborsRegressor(n_neighbors=k, metric='euclidean')
    knn.fit(X_train, y_train)

    # Predict for the left-out sample
    y_pred_all[i] = knn.predict(X_test)[0]

# ---------- 4. Compute C-index over all samples ----------

cidx = c_index(y, y_pred_all)
print("Leave-one-out C-index (sample-wise):", cidx)
```

```python
# ----------------- 1. Load data -----------------
X = np.loadtxt("input.data")          # shape (400, 67)
y = np.loadtxt("output.data")         # shape (400,)
pairs = np.loadtxt("pairs.data",dtype=str)    # shape (400, 2); drug_id, target_id

X = StandardScaler().fit_transform(X)#me

if y.ndim > 1:
    y = y.ravel()

n_samples = X.shape[0]
assert X.shape[0] == y.shape[0] == pairs.shape[0]


# ----------------- 2. C-index -----------------
def c_index(y_true, y_pred):
    """
    Concordance index with 0.5 weight for prediction ties.
    """
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    n = len(y_true)

    concordant = 0
    discordant = 0
    ties = 0

    for i in range(n):
        for j in range(i + 1, n):
            if y_true[i] == y_true[j]:
                # often excluded; here we skip them (no contribution)
                continue
            diff_true = y_true[i] - y_true[j]
            diff_pred = y_pred[i] - y_pred[j]
            prod = diff_true * diff_pred
            if prod > 0:
                concordant += 1
            elif prod < 0:
                discordant += 1
            else:
                ties += 1

    total = concordant + discordant + ties
    if total == 0:
        return np.nan  # or 0.5 by convention
    return (concordant + 0.5 * ties) / total


# ----------------- 3. Helper: generic leave-one-group-out -----------------
def leave_one_group_out_knn(X, y, group_ids, k=10):
    """
    group_ids: 1D array of length n_samples with group labels
               (e.g., drug IDs or target IDs)
    Returns: predictions (n_samples,) from grouped LOO CV.
    """
    unique_groups = np.unique(group_ids)
    y_pred = np.zeros_like(y, dtype=float)

    for g in unique_groups:
        test_idx = (group_ids == g)
        train_idx = ~test_idx

        # Skip groups with no training data (shouldn't happen here)
        if np.sum(train_idx) == 0:
            continue

        X_train, y_train = X[train_idx], y[train_idx]
        X_test = X[test_idx]

        model = KNeighborsRegressor(n_neighbors=k, metric='euclidean')
        model.fit(X_train, y_train)
        y_pred[test_idx] = model.predict(X_test)

    return y_pred


# ----------------- 4. Leave-one-drug-out -----------------
drug_ids = pairs[:, 0]
y_pred_drug = leave_one_group_out_knn(X, y, drug_ids, k=10)
cidx_drug = c_index(y, y_pred_drug)
print("Leave-one-drug-out C-index:", cidx_drug)


# ----------------- 5. Leave-one-target-out -----------------
target_ids = pairs[:, 1]
y_pred_target = leave_one_group_out_knn(X, y, target_ids, k=10)
cidx_target = c_index(y, y_pred_target)
print("Leave-one-target-out C-index:", cidx_target)
```

```python
# ----------------- 1. Load data -----------------
X = np.loadtxt("input.data")          # shape (400, 67)
y = np.loadtxt("output.data")         # shape (400,)
pairs = np.loadtxt("pairs.data",dtype=str)      # shape (400, 2)

if y.ndim > 1:
    y = y.ravel()

n_samples = X.shape[0]
assert X.shape[0] == y.shape[0] == pairs.shape[0], "Shape mismatch"


# ----------------- 2. C-index -----------------
def c_index(y_true, y_pred):
    """
    Concordance index with 0.5 weight for prediction ties.
    Pairs with equal y_true are ignored.
    """
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    n = len(y_true)

    concordant = 0
    discordant = 0
    ties = 0

    for i in range(n):
        for j in range(i + 1, n):
            if y_true[i] == y_true[j]:
                # ignore pairs with equal true values
                continue
            diff_true = y_true[i] - y_true[j]
            diff_pred = y_pred[i] - y_pred[j]
            prod = diff_true * diff_pred
            if prod > 0:
                concordant += 1
            elif prod < 0:
                discordant += 1
            else:
                ties += 1

    total = concordant + discordant + ties
    if total == 0:
        return np.nan
    return (concordant + 0.5 * ties) / total


# ----------------- 3. LOO with type classification -----------------

# Prepare containers to collect true/pred values per type
results = {
    "A": {"y_true": [], "y_pred": []},
    "B": {"y_true": [], "y_pred": []},
    "C": {"y_true": [], "y_pred": []},
    "D": {"y_true": [], "y_pred": []},
}

k = 10

for i in range(n_samples):
    # --- 3.1 Define train/test for this fold ---
    test_idx = i
    train_mask = np.ones(n_samples, dtype=bool)
    train_mask[test_idx] = False

    X_train, y_train = X[train_mask], y[train_mask]
    X_test = X[test_idx].reshape(1, -1)
    y_test = y[test_idx]
    d_test, t_test = pairs[test_idx]  # Di, Tj for the test pair

    # --- 3.2 Determine which drug/targets are present in training ---
    drugs_train = set(pairs[train_mask, 0])
    targets_train = set(pairs[train_mask, 1])

    # presence flags
    d_in_train = d_test in drugs_train
    t_in_train = t_test in targets_train

    # --- 3.3 Determine type for this test pair ---
    # Type A: (Di, Tj) pair not in training, but both Di and Tj appear with other partners
    # Type B: Di unseen, Tj seen
    # Type C: Di seen, Tj unseen
    # Type D: Di unseen and Tj unseen
    #
    # Note: (Di, Tj) is never in training because this is LOO on pairs.
    if d_in_train and t_in_train:
        type_label = "A"
    elif (not d_in_train) and t_in_train:
        type_label = "B"
    elif d_in_train and (not t_in_train):
        type_label = "C"
    else:  # (not d_in_train) and (not t_in_train):
        type_label = "D"

    # --- 3.4 Train model and predict for this test pair ---
    # Edge case: if training set collapses to very few samples, k may be too large.
    # Use min(k, n_train) to avoid errors.
    n_train = X_train.shape[0]
    k_eff = min(k, n_train)

    model = KNeighborsRegressor(n_neighbors=k_eff, metric='euclidean')
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)[0]

    # --- 3.5 Store result under its type ---
    results[type_label]["y_true"].append(y_test)
    results[type_label]["y_pred"].append(y_pred)


# ----------------- 4. Compute C-index per type -----------------
for t in ["A", "B", "C", "D"]:
    y_true_t = np.array(results[t]["y_true"])
    y_pred_t = np.array(results[t]["y_pred"])

    if len(y_true_t) < 2:
        cidx = np.nan
    else:
        cidx = c_index(y_true_t, y_pred_t)

    print(f"Type {t}: n={len(y_true_t)}, C-index={cidx}")
```
